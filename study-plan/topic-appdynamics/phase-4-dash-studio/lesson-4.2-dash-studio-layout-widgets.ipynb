## 1. The Visual Dashboard Editor

The Dash Studio **Visual Dashboard Editor** is a drag-and-drop canvas builder.

### How to add a widget:
1. Click **Edit** mode button (pencil icon)
2. Click the **+** (Add Widget) button in the toolbar
3. Choose the widget type from the gallery
4. The widget appears on the canvas — drag it to position
5. Drag the edges/corners to resize
6. Click the widget to open the **Data panel** (right side)

### Canvas grid:
- The canvas uses a **12-column grid**
- Widgets snap to the grid automatically
- Resize widgets by dragging the resize handle in the bottom-right corner
- Move widgets by clicking and dragging the title bar

### Keyboard shortcuts:
| Shortcut | Action |
|----------|--------|
| `Ctrl+Z` | Undo last change |
| `Ctrl+S` | Save dashboard |
| `Delete` | Remove selected widget |
| `Esc` | Deselect / close panel |

## 2. Widget Configuration Deep Dive

Every widget has a **Data panel** (right sidebar) with two sections:

### Show me data for (data source selector):
Determines *where* data comes from:

| Option | Description |
|--------|-------------|
| **Server Metrics** | Select a metric path for a specific tier or node |
| **Database Metrics** | Select database performance metrics |
| **Custom Query** | Build a query from dropdowns (Application, BT, metric) |
| **ADQL Query** | Write ADQL directly (most powerful — covered in Lesson 4.4) |
| **JMX** | Java MBean metrics |
| **Browser Applications Metrics** | EUM browser metrics |
| **Backends** | Backend/remote service metrics |

### Display settings:
- **Title**: Widget heading
- **Color scheme**: Green/Red/Blue, custom
- **Thresholds**: Color bands based on value ranges
- **Aggregation**: Min, Max, Avg, Current, Sum
- **Time range**: Use dashboard time range, or override per widget

## 3. Dashboard Layout Best Practices

### The F-pattern layout:
Users scan dashboards in an F-shape (top → left). Put the most critical information at the top-left.

```
Row 1: [KEY STAT 1]   [KEY STAT 2]   [KEY STAT 3]   [KEY STAT 4]
       ← Most important metrics at a glance

Row 2: [LARGE TIME SERIES CHART — Response time over time       ]
       ← Trend context: "Is it getting better or worse?"

Row 3: [TABLE: Top BTs      ]  [PIE: Errors by region   ]
       ← Detail: "What specific things need attention?"

Row 4: [HEALTH STATUS       ]  [TABLE: Slow queries      ]
       ← Operational alerts and known issues
```

### Common mistakes to avoid:
- Too many widgets on one page (more than 12 widgets per dashboard is noisy)
- No context — every chart should have a meaningful title and units
- No thresholds — stat widgets without color thresholds don't communicate health at a glance
- No variables — hard-coded dashboards require duplication for every app/tier/environment
